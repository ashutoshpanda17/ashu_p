```python
df_zip = spark.createDataFrame([
    (["Ashu", "Kiran"], [29, 31]),
], ["names", "ages"])

df_zipped = df_zip.withColumn("zipped", arrays_zip("names", "ages"))
df_zipped.show()
```

```python

df_map = df_zip.withColumn("mapped", create_map(lit("Ashu"), lit(29), lit("Kiran"), lit(31)))
df_map.show(truncate=False)
```

```python
df_entries = spark.createDataFrame([
    ([("Ashu", 29), ("Kiran", 31)],),
], ["kv_pairs"])

# Convert tuples into structs, then into map
df_struct = df_entries.withColumn("kv_struct", 
    map_from_entries(expr("transform(kv_pairs, x -> struct(x._1 as key, x._2 as value))"))
)
df_struct.show(truncate=False)
```

5. Display the Empno, Ename, Job, Hiredate, and experience of all Managers.


```python

# Suppose you want to convert 'hiredate' column to DateType
emp = emp.withColumn("hiredate", col("hiredate").cast("date"))
```

```python
# Filter managers and compute experience in years
managers_with_exp = (
    emp.filter(col("job") == "MANAGER")
       .withColumn("experience_years", round(datediff(current_date(), col("hiredate")) / 365.0, 2))
       .select("empno", "ename", "job", "hiredate", "experience_years")
)

managers_with_exp.display()
```
